```python
#pyramid

n=int(input("enter number of rows:"))
print("the pyramid is :")
for i in range(1,n+1):
    for j in range(1,i+1):
        print(i*j,end=" ")
    print("\n")
```

```python
#pattern of * using nested loop
for i in range(0,4):
    for j in range(0,i+1):
        print("*",end=" ")
    print("\n")
for i in range(4,-1,-1):
    for j in range(0,i+1):
        print("*",end=" ")
    print("\n")
```
