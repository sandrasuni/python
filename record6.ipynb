```python
#Leap year

def leap(l,u):
    print("Leap years between ",l, "and", u, ":")
    for i in range (l,u+1):
        if(i%4==0) and (i%100!=0)or(i%400==0):
            print(i)
a=int(input("Enter first year: "))
b=int(input("Enter second year: "))
leap(a,b)
```

```python
#lambda function

sqr=lambda a:a*a
rect=lambda l,b:l*b
tri=lambda b,h:(b*h)/2
while(1):
    print("1.square \n 2.Rectangle \n 3.Triangle \n 4.Exit")
    ch=int(input("Enter your choice:"))
    if(ch==1):
        a=int(input("Enter side:"))
        print("Area of Square:",sqr(a))
    elif(ch==2):
        l=int(input("Enter length:"))
        b=int(input("Enter breadth:"))
        print("Area of Rectangle:",rect(l,b))
    elif(ch==3):
        h=int(input("Enter height:"))
        b=int(input("Enter breadth:"))
        print("Area of Triangle:",tri(b,h))
    else:
        print("Exiting....")
        break
```

```python
#Add ‘ing’ at the end of a given string. If it already ends with ‘ing’, then add ‘ly’.

word=input("Enter a word:")
str1=word[-3:]
if str1=='ing':
    word=word+'ly'
else:
    word=word+'ing'
print("New word:",word)
```

```python
#list of words and return length of longest word

l1=[]
l=list(map(str,input("Enter a list of words with comma separated:").split(",")))
for i in l:
    l1.append(len(i))
print("Length of longest word:",max(l1))
```
