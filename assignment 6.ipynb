```python
#factorial of any number

import math
n=int(input("enter a number:"))
f=1
i=1
while(i<=n):
    f=f*i
    i=i+1
print("factorial is:",f)
```

```python
#fibonacci seies

n=int(input("enter a limit:"))
firstno=1
secondno=0
print("fibonacci series:\n")
print(secondno)
i=1
while(i<=n):
    r=firstno+secondno
    print(r)
    firstno=secondno
    secondno=r
    i=i+1
```
