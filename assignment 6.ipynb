```python
#factorial of any number

import math
n=int(input("enter a number:"))
f=1
i=1
while(i<=n):
    f=f*i
    i=i+1
print("factorial is:",f)
```

```python
#fibonacci seies

n=int(input("enter a limit:"))
firstno=1
secondno=0
print("fibonacci series:\n")
print(secondno)
i=1
while(i<=n):
    r=firstno+secondno
    print(r)
    firstno=secondno
    secondno=r
    i=i+1
```

```python
#greatest common divisor

num1=int(input("enter 1st number:"))
num2=int(input("enter 2nd number:"))
a=num1
b=num2
while num1!=num2:
    if num1>num2:
        num1-=num2
    else:
        num2-=num1
print("GCD of",a,"and",b,"is",num1)
```

```python
#sum of all items on a list

li=list(input("enter numbers:").split(","))
ln=len(li)
a=0
i=0
for i in range(ln):
    b=int(li[i])
    a=a+b
print("Sum of the elements in the list",li,"is",a)
```

```python
#Average of Tuple

t=tuple(input("Enter the numbers:").split(","))
print("Tuple : ",t)
ln=len(t)
c=0
for i in range(ln):
    c=int(t[i])+c
    avg=c/ln
print("Average values of the number in a given tuple",avg)
```

```python
#frequency of characters in a string

s=input("Enter a string :")
d={}
for i in s:
    if i in d:
        d[i]=d[i]+1
    else:
        d[i]=1
for keys,value in d.items():
    print(keys,"is repeated",value,"time")
        
```
