```python
#first and last characters exchanged
s=str(input("enter a string:"))
a=s[0]
b=s[-1]
print("the resultant string=",b+s[1:-1]+a)
```

```python
#string made of 1st 2 and last 2 characters
s=str(input("enter a string:"))
a=s[0]
b=s[-1]
c=s[1]
d=s[-2]
print("the resultant string =",a+c+d+b)
```
