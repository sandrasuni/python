```python
#positive,negative or zero

n=int(input("enter a number:"))
if(n>0):
    print("given number is positive")
elif(n==0):
    print("given number is zero")
else:
    print("given number is negative")
```

```python
#greatest of 3 numbers

a=int(input("enter the 1st no:"))
b=int(input("enter the 2nd no:"))
c=int(input("enter the 3rd no:"))
if(a>b and a>c):
    print("the greatest no among the 3 numbers is:",a)
elif(b>a and b>c):
    print("the greatest no among the 3 numbers is:",b)
else:
    print("the greatest no among the 3 numbers is:",c)
```

```python
#calculator

a=int(input("enter the 1st no:"))
b=int(input("enter the 2nd no:"))
print("1.Addition")
print("2.Subtraction")
print("3.Multiplication")
print("4.Division")
choice=int(input("select option from 1,2,3,4:"))
if(choice==1):
    print("the sum is:",a+b)
elif(choice==2):
    print("the subtracted value is:",a-b)
elif(choice==3):
    print("the multiplied value is:",a*b)
elif(choice==4):
    print("the remainder is:",a/b)
else:
    print("wrong choice")
```
