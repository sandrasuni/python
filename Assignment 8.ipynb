```python
#reverse a tuple

t=tuple(input("Enter the elements:").split(","))
print("Tuple before reversing: ",t)
t=tuple(reversed(t))
print("reversed tuple is :",t)
```

```python
#3rd element and 3rd element from last of a tuple

t=tuple(input("Enter the elements:").split(","))
print("Tuple: ",t)
i=t[2]
print("third element of the tuple is :",i)
j=t[-3]
print("3rd element from last of tuple:",j)
```

```python
#remove an item from a tuple

t=tuple(input("Enter the elements:").split(","))
l=list(t)
print("Tuple: ",t)
a=input("the element to be deleted:")
for x in t:
    if a in x:
        l.remove(x)
        r=tuple(l)
print(r)
```
