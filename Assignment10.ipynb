```python
#list comprehension
#positive list of numbers

print("positive list of numbers \n")
n=input("Enter numbers:").split(",")
a=map(int,n)
new1=[x for x in a if x>0]
print("positive integers from the given list are:",new1)
print("\n")

#square of n numbers

print("square of n numbers \n")
new2=[x**2 for x in new1]
print("square of numbers in the list is",new2)
print("\n")

#list of vowels

print("list of vowels \n")
vow=input("enter the word:")
vow1=[x for x in vow]
vowel=['a','e','i','o','u','A','E','I','O','U']
vow2=[x for x in vow1 if x in vowel]
print("vowels in",vow,"are",vow2)
print("\n")

#ordinal values

print("ordinal values \n")
ordinal=[ord(x) for x in vow1]
print("ordinal values of",vow1,"are:",ordinal)
```

```python
#list of integers value greater than 100 store 'over
values=[]
while True:
    value=input("enter an integer('enter stop to stop'):")
    if value=='stop':
        break    
    elif int(value)>100:
        values.append('over')
    else:
        values.append(int(value))
print(values)
```

```python
#Print out all colors from color-list1 not contained in color-list2

c1=set(input('Enter colors for first list: ').split(','))
c2=set(input('Enter colors for second list: ').split(','))
print('Difference of ',c1,' and ',c2,': ')
print(c1.difference(c2))
```

```python
#check two list
lst1=set(input('Enter first list with comma separated: ').split(','))
lst2=set(input('Enter second list with comma separated: ').split(','))
sum1 = str(0)
sum2 = str(0)
if len(lst1) == len(lst2):
    print("both list are of equal length")
else:
    print("Two lists have unequal length")
for x in lst1:
    sum1 = sum1 + x
for x in lst2:
    sum2 = sum2 + x
if sum1 == sum2:
    a = "equal"  
else:
    a = "not equal"  
print("sum of two list are ", a)  
for x in lst1:
    for y in lst2:
        if x == y:
            print(y," occurs in both list")
```

```python
#Store a list of first names.Count the occurrences of ‘a’ within the list.
names=set(input('Enter list comma separated: ').split(','))
print(names)
count=0
for name in names:
    count+=name.count('a')
print("occurence of a in",names," is:",count)
```
