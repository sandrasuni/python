```python
#dictionary ascending and descending order

dic={}
n=int(input("no of elements:"))   
for i in range(n):
    f=input("enter the name:")
    kg=int(input("enter the quantity:"))
    dic[f]=kg
print("Dictionary:",dic)
l=list(dic.items())
l.sort()
ac=dict(l)
print("Dictionary in ascending order:",ac)
r=list(dic.items())
r.sort(reverse=True)
dc=dict(r)
print("Dictionary in descending order:",dc)
```

```python
dic1={}
n=int(input("no of elements:"))   
for i in range(n):
    f=input("enter the name of fruits:")
    kg=int(input("enter the quantity:"))
    dic1[f]=kg
print("1st Dictionary:",dic1)
dic2={}
n=int(input("no of elements:"))   
for i in range(n):
    f=input("enter the name of flower:")
    kg=int(input("enter the quantity:"))
    dic2[f]=kg
print("2nd Dictionary:",dic2)
d3=dic1
dic1.update(dic2)
print("merge dictionary is",d3)
```
