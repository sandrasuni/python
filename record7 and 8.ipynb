```python
#Create modules Rectangle, Circle.

import rectangle
import cle
while(1):
    print("1.Rectangle\n2. Circle\n3. Exit")
    ch=int(input("Enter your choice:"))
    if ch==1:
        l=int(input("Enter length"))
        b = int(input("Enter breadth"))
        rectangle.rectarea(l,b)
        rectangle.rectperi(l,b)
    elif ch==2:
        r = int(input("Enter radius"))
        cle.circlearea(r)
        cle.circleperi(r)
    else:
        print("Exiting.....")
        break
```

```python
#package graphics with modules rectangle, circle and subpackage graphics3D with modules cuboid and sphere.

from graphics import rectangle
from graphics import circle
from graphics.graphics3d import cuboid
from graphics.graphics3d import sphere
while(1):
    print("1. Rectangle\n2. Circle\n3. Cuboid\n 4. Sphere\n 5. Exit")
    ch=int(input("Enter your choice: "))
    if ch==1:
        l=int(input("Enter length: "))
        b=int(input("Enter breadth: "))
        rectangle.rectarea(l,b)
        rectangle.rectperi(l,b)
    elif ch==2:
        r=int(input("Enter radius: "))
        circle.circlearea(r)
        circle.circleperi(r)
    elif ch==3:
        l = int(input("Enter length: "))
        b = int(input("Enter breadth: "))
        h= int(input("Enter height: "))
        cuboid.cuboidarea(l,b,h)
        cuboid.cuboidperi(l,b,h)
    elif ch==4:
        r=int(input("Enter radius: "))
        sphere.spherearea(r)
        sphere.sphereperi(r)
    else:
        print("Exiting...")
        break
```
